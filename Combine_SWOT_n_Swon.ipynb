```python
import numpy as np
import xarray as xr

D   = '/SCRATCH/n23nguye/mapping_2023_2025'
OUT = f'{D}/alongtrack_SWOTnSwon_real_2023-01-01_2025-07-31.nc'

a = xr.open_dataset(f'{D}/alongtrack_swon_real_2023-01-01_2025-07-31.nc')['sla']
b = xr.open_dataset(f'{D}/alongtrack_SWOT_real_2023-01-01_2025-07-31.nc')['sla']

a, b = xr.align(a, b, join='exact')          # fails loudly if grids differ

# mean where both, the single value where only one -> NaN only where neither
comb = xr.concat([a, b], dim='sat').mean(dim='sat', skipna=True)

comb.name = 'sla'
comb.to_netcdf(OUT)

print('swon     %.4f%%' % (100 * float(np.isfinite(a).mean())))
print('SWOT     %.4f%%' % (100 * float(np.isfinite(b).mean())))
print('combined %.4f%%' % (100 * float(np.isfinite(comb).mean())))
print('->', OUT)
```
