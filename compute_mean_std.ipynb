```python
import xarray as xr
F = "/Odyssey/public/glorys/reanalysis/glorys12_2010_2019_daily_sla_8th_gridded_from_alongtrack_float32.nc"
v = xr.open_dataset(F, chunks={"time": 50})["tgt"]          # tgt = gap-free GLORYS (not obs)
for name, sl in {"train": slice("2010-01-01", "2017-10-31"), "val": slice("2017-12-01", "2018-11-30")}.items():
    x = v.sel(time=sl).isel(time=slice(None, None, 5))       # every 5th day: fast, accurate enough
    print(name, float(x.mean()), float(x.std()))
```
